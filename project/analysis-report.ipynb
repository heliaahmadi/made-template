# Final Analysis Report on Energy Consumption and CO2 Emissions in London Boroughs

## Introduction

### Project Overview
This research undertakes an in-depth study of energy use and CO2 emission across the London boroughs in light of global climate challenges. It has emphasized the need for acting on urban centers to lead towards sustainable development.

### Objectives
The main objectives of this work are :
- To identify patterns and trends in energy use and CO2 emissions; 
- To provide information for policy action to maintain and bring environmental benefits with the gained insights.

### Significance
Such an analysis could be instrumental to urban and environmental planning for the accomplishment of sustainable development goals in London by providing relevant data-driven insights.

## Used Data

### Source and Selection
Data was sourced from the [London Datastore](https://data.london.gov.uk/), chosen for its comprehensive coverage of energy and emission metrics. This source was selected due to its relevance and the breadth of information it offers for urban studies.

- Energy Consumption by Borough: [Download Dataset](https://data.london.gov.uk/download/leggi/2d6ee3f1-e928-48a9-8eab-01748c65ac6f/energy-consumption-borough-leggi.csv)
- CO2 Emissions by Borough: [Download Dataset](https://data.london.gov.uk/download/leggi/fcf8a0a3-2051-484a-ba9a-5c8bc2268a3e/co2-emissions-borough-leggi.csv)

### Data Structure
The datasets hold a column structure containing boroughs, years, and metrics of energy
This includes consumption in MWh and CO2 emissions in metric tons, which provides a granular view over time.

### Data Licensing and Compliance
The dataset is licensed under the UK Open Government Licence, which allows for the modification and
sharing of the data in transparent and ethical use within public domain research.

### CO2 Emissions Analysis
The table of long-term mean CO2 emissions across London through its boroughs indicates that there are distinct trends. Notably, Westminster shows an approximate continuous decrease from 2010 to 2017. There might be any number of factors driving this trend, from local policy to changes in broad energy sources and patterns of use. Many other cases, such as Camden and the City of London, give the impression of fluctuation but no definite downward trend; perhaps it is here, therefore, that the most concentrated environmental policy efforts need to be focused.

#### Key Observations:

Westminster's reducing emissions pattern could be taken as an important case study for other boroughs since it has a steady pattern.
Variability across Camden and the City of London could therefore reflect the diverse nature of economic activities and urban dynamics that may influence emissions.

### Energy Consumption Analysis
This energy consumption chart is very different from the emissions data, indicating that in most of the boroughs, the level of consumption is relatively flat with slight declines at the end of the period. Specifically, in Kensington and Chelsea, there is an erratic pattern, which may be correlated to either changes in residential or commercial uses or anomalies that may need further probing in the data.

#### Key Observations:

It shows stability across most London boroughs, indicating that a reduction in CO2 emission for some areas might not be a case of reduced energy consumption.
What could influence the profile of Kensington and Chelsea is demography, new buildings, or another factor driving change in the pattern of energy demand.
Comparative Analysis
A comparative review of the two charts thereby indicates that reductions in CO2 emissions do not always run parallel with decreases in energy use. This quite clearly would suggest energy efficiency improvements or cleaner energy source displacements. The data also offers an opportunity, therefore, to examine the role of technologies such as renewable energy adoption or energy-saving schemes.

## Conclusions

### Summary of Insights
Results depict that few of the boroughs are significantly on higher energy consumption and related emissions. A few of the findings are: 

- Boroughs with higher industrial activities tend to have higher energy consumption and CO2 emissions.
- Emissions have notably come down in those boroughs where renewable sources of energy have replaced traditional sources.

### Uncertainties and Limitations
- Incompleteness of data for some years may affect the trend analysis.
- All data entries are assumed accurate as in the source data for analysis.

### Future Work
This would point to further research in the future, probably incorporating more variables like economic factors and demographic data to add depth to the analysis and make more nuanced policy recommendations.

### Policy Implications
In consideration of the findings, we recommend that local authorities weigh options with renewable energy technologies and initiatives for reduced energy use and emissions in the high-impact boroughs.
